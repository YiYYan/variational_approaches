```python
import numpy as np
import matplotlib.pyplot as plt
from matplotlib import cm
plt.rcParams['xtick.direction']='in'
plt.rcParams['ytick.direction']='in'
plt.rcParams['font.family']='times new roman'
plt.rcParams['mathtext.fontset']='stix'
plt.rcParams['font.size']=13
#plt.rcParams['font.size']=11
```

```python
def plot_dyn(xx,yy,axs,ls,leg,lo):
    axs.plot(xx,yy,ls,label=leg)
    axs.set_box_aspect(0.618)
    axs.legend(loc=lo,frameon=False,ncol=1,fontsize=11)
    axs.set_xlim(0,300)
    #axs.set_ylim(0,0.5)
```

```python
def plot_log(xx,yy,axs,ls,leg,lo):
    axs.semilogy(xx,yy,ls,label=leg)
    axs.set_box_aspect(0.618)
    axs.legend(loc=lo,frameon=False,ncol=1,fontsize=11)
    axs.set_xlim(0,300)
```

```python
def plot_fun(xx,yy,axs,ls,leg,lo):
    axs.plot(xx,yy,ls,label=leg)
    axs.set_box_aspect(0.618)
    axs.legend(loc=lo,frameon=False)
    axs.set_xlim(0.0,0.5)
    #axs.set_ylim(0,6.8)
```

```python
def jwfun(alpha,x):
    return 2*alpha*x*np.exp(-x**2)
```

```python
fig3,ax3=plt.subplots(nrows=1, ncols=3, figsize=(12,3.2), layout="constrained")
plot_dyn(ttc3,np.real(szlistc3[0]),ax3[0],'-','$M=2$','best')
plot_dyn(ttc,np.real(szlistc[0]),ax3[0],'--','$M=5$','best')
plot_dyn(ttc2,np.real(szlistc2[0]),ax3[0],':','$M=7$','best')

plot_dyn(ttc3,np.real(gzzfunc3[200]),ax3[1],'-','$M=2$','best')
plot_dyn(ttc,np.real(gzzfunc[200]),ax3[1],'--','$M=5$','best')
plot_dyn(ttc2,np.real(gzzfunc2[200]),ax3[1],':','$M=7$','best')

plot_fun(wkc3,0.25*jwfun(0.5,wkc3)*np.real(phlist2c3[-1]),ax3[2],'-','$M=2$','best')
plot_fun(wkc,0.25*jwfun(0.5,wkc)*np.real(phlist2c[-1]),ax3[2],'--','$M=5$','best')
plot_fun(wkc2,0.25*jwfun(0.5,wkc2)*np.real(phlist2c2[-1]),ax3[2],':','$M=7$','best')

ax3[0].text(10,0.95,r'(a)')
ax3[1].text(10,0.95,r'(b)')
ax3[2].text(0.03,73,r'(c)')
for jj in range(0,2):
    ax3[jj].set_xlabel(r'$t$ (units of $\omega_c^{-1}$)')

ax3[0].set_ylabel(r'$\langle\sigma_z(t)\rangle$')
ax3[1].set_ylabel(r'$\langle\sigma_z(t)\sigma_z(t^\prime)\rangle$')
ax3[2].set_ylabel(r'$I_q(t)$')
ax3[2].set_xlabel(r'$\omega_q$ (units of $\omega_c$)')
#plt.savefig("D:/data/smatpi/FigSnewc.pdf", format='pdf', dpi=300)
```
